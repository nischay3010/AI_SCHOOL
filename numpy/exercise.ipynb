Valid Sudoku

```python
import numpy as np
```

```python
s=np.array([
    [5,3,4,6,7,8,9,1,2],
    [6,7,2,1,9,5,3,4,8],
    [1,9,8,3,4,2,5,6,7],
    
    [8,5,9,7,6,1,4,2,3],
    [4,2,6,8,5,3,7,9,1],
    [7,1,3,9,2,4,8,5,6],
    
    [9,6,1,5,3,7,2,8,4],
    [2,8,7,4,1,9,6,3,5],
    [3,4,5,2,8,6,1,7,9]
])
```

```python
sum_column=np.sum(s,axis=0)
sum_row=np.sum(s,axis=1)

if np.array_equal(sum_column,sum_row):
    print("Valid Sudoku")
     
```

```python
print(sum_column)
```

```python
sum_column[sum_column==45]
```

```python
if np.all(sum_column==45) & np.all(sum_row==45):
    print("Valid Sudoku")
else:
    print("Not valid sudoku")
```
