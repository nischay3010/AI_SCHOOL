```python
prices=np.array([60000,800,2000,15000,3000,4000])
monthly_sales=np.array([
    [8,10,12,15],
    [60,70,85,90],
    [25,30,35,40],
    [10,12,15,10],
    [30,35,40,45],
    [12,10,8,6]
])
```

```python
monthly_revenue=monthly_sales*prices[:,np.newaxis]
print(monthly_revenue)
```

```python
product_revenue=np.sum(monthly_revenue,axis=1)
print(product_revenue)
```

```python
revenue_by_month=np.sum(monthly_revenue,axis=0)
print(revenue_by_month)
```

```python
top_index=np.argmax(product_revenue)
print(top_index)
```
